### Diseño y planificación
**La página de overview** tendrá como objetivo presentar una visión general del desempeño comercial de Andes Retail Group durante 2024 y 2025. Esta vista permitirá identificar rápidamente la evolución de los ingresos, el comportamiento de la rentabilidad y los principales elementos que aportan valor al negocio. Mostrando cómo evolucionaron los ingresos, qué segmentos aportan más al negocio y qué categorías de producto tienen mayor impacto.    
**La página de detalle** tendrá como objetivo profundizar en las causas del comportamiento de los ingresos de Andes Retail Group durante 2024 y 2025. Esta vista permitirá analizar con mayor precisión qué factores explican las subidas, caídas o estabilidad de los ingresos. Ayudando a responder qué segmentos y productos aportan mayor valor, si existen diferencias relevantes entre ubicaciones geográficas, qué patrones temporales se presentan a lo largo del año y dónde podrían existir oportunidades de mejora comercial.

#### Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?

**KPIs**
- `Ingresos:` Ventas totales. Permite medir el ingreso total generado por el negocio en el perido analizado, nos permite evaluar el tamaño comercial de la operación y comparar entre 2024 y 2025
- `Costos:` Costo total. Permite conocer cuánto le cuesta al negocio generar esas ventas, para analizar la eficiencia comercial este KPI nos permite entender si el crecimiento de ingresos implicaría un aumento proporcional en los costos.
- `Utilidad:` Utilidad bruta y margen bruto. Muestra la ganancia, KPI clave para evaluar que tan rentable es el negocio
- `Variación 2024 vs 2025:` Muestra el porcentaje de variación positivo o negativo en las ventas del 2024 contra el 2025

**Gráficos**
- `Gráfico de tiempo:` Patrón estacional
  - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?. Este grafico muestra la evolución a lo largo de las estaciones, comparando cada año, de esta forma se identificará exactamente cuando caen o suben las ventas.
- `Gráfico comparación region / Ingresos:` Región / Ventas totales
  - ¿Existen diferencias relevantes entre países o regiones?. Este grafico nos responde la pregunta de la relevancia por region
- `Gráfico comparación Ingresos vs clientes/productos:` 
    - Ingresos comparando segmento de clientes / productos    
  - ¿Qué segmentos de clientes aportan mayor ingreso y rentabilidad? y ¿Qué categorías de producto tienen mayor impacto en el negocio?, con este grafico podemos comparar los ingresos por segmento de clientes y productos
- `Gráfico comparación Margen vs clientes/productos:` 
    - Rentabildad comparando segmento de clientes / productos 
  - ¿Qué segmentos de clientes aportan mayor ingreso y rentabilidad? y ¿Qué categorías de producto tienen mayor impacto en el negocio?, con esto grafico podemos comparar la rentabilidad por segmento de clientes y productos
- `Gráfico comparación ingresos / país:`  Ventas totales /  País
  - ¿Existen diferencias relevantes entre países o regiones?. Este grafico nos responde la pregunta de la relevancia por país

#### 🖥️ Vista General (Overview)

**S (Situación):**  
Panorama comercial de Andes Retail Group, que compara el período 2024 vs 2025
Las ventas totales son de 5.5 millones (2.9 mill. para 2024 y 2.7 mill. para 2025), con una rentabilidad total de 3.6 millones que es un margen bruto de 35.1 puntos porcentuales, el margen es el mismo para cada año.
Los ingresos están definidos por el tipo de cliente y no por las diferentes categorías de productos, asimismo la rentabilidad promedio de los productos se mantiene oscilando al rededor del 35%, Perú tiene la mayor participación en los ingresos con 2.2 mill., seguido de chile con 2 mill. y Colombia con 1.3 mill.

**C (Complicación):**  
La variación anual 2024 vs 2025 fue de 6.8 puntos porcentuales negativos en las ventas, los cuales representan una diferencia de $194,774

**Q (Pregunta):**   
¿Por qué ocurre esta caída?, más específicamente, cuándo y cómo ocurre

**A (Respuesta):**  
Se observa un patrón estacional, en el cual se observa el verano como la temporada alta de ventas y el invierno es la temporada baja, sin embargo la variación negativa en las ventas ocurrió en mayor medida en la temporada alta de verano.

#### 🔎 Vista Detalle

**S (Situación):**  
Se observa un patrón estacional, temporada alta en verano y las ventas caen drásticamente en invierno (Junio - Agosto), baja la cantidad de unidades vendidas y baja el monto del ticket promedio.    
Las ventas altas superan a las ventas bajas por tres estaciones, pero en invierno las ventas altas bajan drásticamente incluso son menores a las ventas bajas, las cuales se mantuvieron constantes.

**C (Complicación):**  
Se espera que las ventas bajen en invierno, sin embargo también bajaron en verano de `$1,222,301` a `$1,019,418` un 16.5% con respecto al año anterior

**Q (Pregunta):**  
Se sabe que las ventas no fueron las esperadas en la temporada de verano ¿por qué ocurrió esto?, ¿Ocurrió en todos los países y regiones o fue segmentado? ¿Existe algún patrón?

**A (Respuesta):**  
Esto ocurrio por que el segmento de ventas altas no siguió el comportamiento esperado, particularmente en las "ventas altas" de la zona sur y centro, aunque afecta a todos los países, se ve más afectado Perú porque tiene la mayor participación de ingresos.

**📊 Actualización desempeño comercial**

Buenos días Equipo, les comparto los hallazgos clave del análisis de ventas 2024 vs 2025.

**Lo que encontramos:** Las ventas cayeron -6.8% YoY (~\$194,774 menos que 2024). El margen bruto se mantiene estable en 35.1% en ambos años — el problema no es rentabilidad, es volumen.    
**Por qué está pasando:** La caída no es la estacionalidad esperada (invierno siempre es temporada baja). El golpe real fue en verano, donde las ventas altas bajaron 16.5% (`$1.22M → $1.02M` vs 2024) — un quiebre fuera del patrón normal.    
**Dónde se concentra:** El problema es específico, no generalizado:
- Afecta principalmente al segmento de ventas altas, no a ventas bajas (que se mantuvieron estables)
- Se concentra en regiones Sur y Centro
- Impacta más a Perú, por ser el país con mayor participación de ingresos (2.2M)


**Impacto en negocio:** Como Perú concentra la mayor parte del ingreso total, esta caída regional tiene efecto directo y desproporcionado en el resultado consolidado del grupo.

**Acciones recomendadas:**
1. 🎯 `Priorizar una revisión comercial en Perú y Colombia`, específicamente en el segmento de ventas altas de zonas Sur/Centro, durante verano
2. 🔍 `Investigar causa raíz`: ¿fue pricing, disponibilidad de producto, competencia, o algo puntual del período?
3. 📅 `Monitorear si el patrón se repite en el próximo verano` — si es estructural y no puntual, necesita plan de mitigación antes de la temporada
